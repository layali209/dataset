```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Toy dataset
toy_dataset = [
    ['Skirt', 'Sneakers', 'Scarf', 'Pants', 'Hat'],
    ['Sunglasses', 'Skirt', 'Sneakers', 'Pants', 'Hat'],
    ['Dress', 'Sandals', 'Scarf', 'Pants', 'Heels'],
    ['Dress', 'Necklace', 'Earrings', 'Scarf', 'Hat', 'Heels', 'Hat'],
    ['Earrings', 'Skirt', 'Skirt', 'Scarf', 'Shirt', 'Pants']
]

```

```python
te = TransactionEncoder()
te_ary = te.fit(toy_dataset).transform(toy_dataset)
df_toy = pd.DataFrame(te_ary, columns=te.columns_)
df_toy
```

```python
frequent_itemsets = apriori(df_toy, min_support=0.4, use_colnames=True)
frequent_itemsets
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python

df = pd.read_csv('c:/Users/layel/Downloads/Market_Basket_Optimisation.csv')
df.head()
```

```python
transactions = df.values.tolist()
```

```python

transactions = [[item for item in row if pd.notna(item)] for row in df.values]

transactions[:5]
```

```python
all_items = sorted(set(item for transaction in transactions for item in transaction))

encoded_data = []
for transaction in transactions:
    encoded_data.append([1 if item in transaction else 0 for item in all_items])
df_encoded = pd.DataFrame(encoded_data, columns=all_items)
df_encoded.head()
```

```python
min_support = 0.05  # 5% of transactions
frequent_itemsets = []

for item in all_items:
    support = df_encoded[item].mean()
    if support >= min_support:
        frequent_itemsets.append((item, support))


freq_df = pd.DataFrame(frequent_itemsets, columns=['item', 'support']).sort_values(by='support', ascending=False)
freq_df
```

```python
from itertools import combinations

rules = []

for transaction in transactions:
    for combo in combinations(transaction, 2):  # pairs of items
        rules.append(combo)

from collections import Counter
pair_counts = Counter(rules)

n_trans = len(transactions)

rules_list = []
for (a, b), count in pair_counts.items():
    support = count / n_trans
    confidence = count / df_encoded[a].sum()  
    rules_list.append({'antecedent': a, 'consequent': b, 'support': support, 'confidence': confidence})

rules_df = pd.DataFrame(rules_list)
rules_df.sort_values(by='confidence', ascending=False).head(10)
```
